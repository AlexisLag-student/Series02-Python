## Mathematical definition of a sine wave

A sinusoidal signal is defined as:

$$ s(t) = A \sin(2 \pi f t + \varphi) $$

where $A$ is the amplitude, $f$ the frequency in hertz (Hz), $t$ the time in
seconds (s), and $\varphi$ the phase in radians (rad).

The period is $T = 1/f$ and the angular frequency is $\omega = 2 \pi f$ (rad/s).

In this first draft, $A = 1$, $f = 1$ Hz and $\varphi = 0$, so $s(t) = \sin(2 \pi t)$:
exactly one period over $t \in [0, 1]$.
