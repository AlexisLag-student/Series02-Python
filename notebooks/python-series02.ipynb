## Mathematical definition of a sine wave

A sinusoidal signal is defined as:

$$ s(t) = A \sin(2 \pi f t + \varphi) $$

where $A$ is the amplitude, $f$ the frequency in hertz (Hz), $t$ the time in
seconds (s), and $\varphi$ the phase in radians (rad).

The period is $T = 1/f$ and the angular frequency is $\omega = 2 \pi f$ (rad/s).

In this first draft, $A = 1$, $f = 1$ Hz and $\varphi = 0$, so $s(t) = \sin(2 \pi t)$:
exactly one period over $t \in [0, 1]$.

## Cosine wave

The cosine is a sine shifted by a quarter of a period:

$$ c(t) = A \cos(2 \pi f t) = A \sin\left(2 \pi f t + \frac{\pi}{2}\right) $$

So a cosine is a sinusoid with phase $\varphi = \pi/2$.

```python
import numpy as np
import matplotlib.pyplot as plt

DURATION_S = 1.0
N_SAMPLES = 1000
FREQ_LOW_HZ = 1.0
FREQ_HIGH_HZ = 2.0
AMPLITUDE = 1.0
```

## A reusable sinusoid function

Sine and cosine are both special cases of the general sinusoid
$x(t) = A \sin(2 \pi f t + \varphi)$ (the cosine is the case $\varphi = \pi/2$).
Instead of writing a new formula for each signal, we define one function
whose inputs are the parameters $A$, $f$ and $\varphi$, and use it to build
two signals with different frequencies.

```python
def sinusoid(time_s, amplitude=1.0, frequency_hz=1.0, phase_rad=0.0):
    """Compute a sinusoidal signal x(t) = A sin(2 pi f t + phi).

    Parameters
    ----------
    time_s : numpy.ndarray
        Time instants, in seconds.
    amplitude : float, optional
        Peak value of the signal. Default is 1.0.
    frequency_hz : float, optional
        Frequency, in hertz. Default is 1.0.
    phase_rad : float, optional
        Phase at t = 0, in radians. Use pi/2 to get a cosine. Default is 0.0.

    Returns
    -------
    x : numpy.ndarray
        Signal values, with the same shape as `time_s`.
    """
    x = amplitude * np.sin(2 * np.pi * frequency_hz * time_s + phase_rad)
    return x
```

```python
time_s = np.linspace(0, DURATION_S, N_SAMPLES)
signal_low = sinusoid(time_s, AMPLITUDE, FREQ_LOW_HZ)
signal_high = sinusoid(time_s, AMPLITUDE, FREQ_HIGH_HZ)

fig, ax = plt.subplots(figsize=(6, 4))
# Labels are built from the constants so the legend stays correct if frequencies change
ax.plot(time_s, signal_low, label=f"Sinus 1")
# Dashed line keeps the two curves distinguishable when printed in grayscale
ax.plot(time_s, signal_high, label=f"Sinus 2")
ax.set_xlabel("Time")
ax.set_ylabel("Amplitude")
ax.set_title("Sinusoidal function")
ax.grid(True)
ax.legend(loc="upper right")

fig.savefig("sinusoidal_function.pdf")
plt.show()
```
