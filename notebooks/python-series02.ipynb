## Mathematical definition of a sine wave

A sinusoidal signal is defined as:

$$ s(t) = A \sin(2 \pi f t + \varphi) $$

where $A$ is the amplitude, $f$ the frequency in hertz (Hz), $t$ the time in
seconds (s), and $\varphi$ the phase in radians (rad).

The period is $T = 1/f$ and the angular frequency is $\omega = 2 \pi f$ (rad/s).

In this first draft, $A = 1$, $f = 1$ Hz and $\varphi = 0$, so $s(t) = \sin(2 \pi t)$:
exactly one period over $t \in [0, 1]$.

## Cosine wave

The cosine is a sine shifted by a quarter of a period:

$$ c(t) = A \cos(2 \pi f t) = A \sin\left(2 \pi f t + \frac{\pi}{2}\right) $$

So a cosine is a sinusoid with phase $\varphi = \pi/2$.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)
s = np.sin(2 * np.pi * t)
c = np.cos(2 * np.pi * t)

fig, ax1 = plt.subplots()

line_sin, = ax1.plot(t, s, color="tab:blue", label="Sinus")
ax1.set_xlabel("Time")
ax1.set_ylabel("Amplitude")
ax1.set_title("Sinusoidal function")
ax1.grid(True)

ax2 = ax1.twinx()
line_cos, = ax2.plot(t, c, color="tab:orange", label="Cosinus")
ax2.legend(handles=[line_sin, line_cos], loc="lower left")

fig.savefig("sinusoidal_function.png")
plt.show()
```
